1.7 Name Numerology Value Calculator

```
name = input()
total = 0
for ch in name.lower():
    total = total + (ord(ch) - 96)
print(total)
```

1.9 Average Word Length Calculator

```
S = input("String: ")
words = S.split()
nwords = len(words)
total_length = 0
for i in words:
  total_length = total_length + len(i)
avg = total_length/nwords
print(avg)
```

1.10 Future Value Calculator

```
principal = float(input())
rate = float(input())
years = int(input())
print("Year Value")
print("----------------")
for year in range(0, years + 1):
    value = principal * (1 + rate) ** year
    print(f"{year} ${value:.2f}")
```
