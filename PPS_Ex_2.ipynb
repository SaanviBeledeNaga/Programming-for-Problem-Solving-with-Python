2.1 Art Gallery Inventory Checker

```python
paintings = ["Starry Night", "The Mona Lisa", "The Last Supper", "The Scream", "Girl with a Pearl Earring"]
choice = input("Enter the painting you want to buy: ")
count = 0
for p in paintings:
    if p == choice:
        count = count + 1
if count > 0:
    print(f'Yes, we have "{choice}" available for purchase!')
else:
    print(f'Sorry, we don\'t have "{choice}" right now.')
```

2.2 Shoe Store

```python
shoes = input().split(", ")
add_item = input()
remove_item = input()
request = input()
shoes.append(add_item)
shoes = [shoe for shoe in shoes if shoe != remove_item]
print("Final list of shoes:", shoes)
count = 0
for shoe in shoes:
    if shoe == request:
        count = count + 1
if count > 0:
    print(f"Yes, {request} are available in the store.")
else:
    print(f"Sorry, {request} are not available in the store.")
```

2.3 Currency Exchange Office

```python
available = input().split(", ")
to_order = input().split(", ")
request = input()
count = 0
for currency in available:
    if currency == request:
        count = count + 1
if count > 0:
    print(f"Yes, {request} are available.")
    for i in range(len(available)):
        if available[i] == request:
            available.pop(i)
    to_order.append(request)
else:
    print(f"Sorry, {request} are not available.")
    to_order.append(request)
print("Updated available currencies:", available)
print("Updated currencies to order:", to_order)
```

2.5 Ordering a T-shirt Online

```python
tshirt = input().split(", ")
new_text = input()
try:
    index = tshirt.index("add your text here")
    print("Position of 'add your text here' in the list:", index)
    tshirt[index] = new_text
    print("Updated T-shirt features:", tshirt)
except ValueError:
    print(f"Error: 'add your text here' was not found in the T-shirt features: {tshirt}.")
    print(f"Therefore, '{new_text}' could not replace it.")
    print(f"Current T-shirt features remain: {tshirt}")
```

2.8 Word Search

```python
words = ["cards", "park", "pets", "football", "golf", "crosswords", "toys", "exercise",
         "hobbies", "riding", "biking", "games", "reading", "movies", "walking", "concerts"]
print(f"Word search with {len(words)} words")
five_letter = [w for w in words if len(w) == 5]
print("Words with 5 letters:", five_letter)
print("Number of 5-letter words:", len(five_letter))
print("Words with less than 5 letters:")
for i, w in enumerate(words):
    if len(w) < 5:
        print(f"- '{w}' (position {i}) has {len(w)} characters.")
print("Words with more than 8 characters:")
for i, w in enumerate(words):
    if len(w) > 8:
        print(f"- '{w}' (position {i}) has {len(w)} characters.")
half = len(words) // 2
print("Words in the second half (different from 7 characters):")
for i in range(half, len(words)):
    if len(words[i]) != 7:
        print(f"- '{words[i]}' (position {i}) = {len(words[i])} characters")
quarter = len(words) // 4
print("4-letter words in the first fourth of the list:")
for i in range(quarter):
    if len(words[i]) == 4:
        print(f"- '{words[i]}' (position {i})")
```

2.9 Spelling Competition

```python
msk_words = ["ankle", "patella", "rib", "femur", "sternocleidomastoid", "tendon", "sternum",
             "abdominal external oblique", "muscle", "scapula", "radius", "bone",
             "vertebra", "ligament", "ulna", "skull", "clavicle"]
print("Total number of words to learn:", len(msk_words))
print("Length of each word:")
for w in msk_words:
    print(f"{w}: {len(w)}")
short = ["leg"]
intermediate = ["cartilage"]
long = ["pectoralis major"]
for w in msk_words:
    if len(w) <= 6:
        short.append(w)
    elif 7 <= len(w) <= 9:
        intermediate.append(w)
    else:
        long.append(w)
print("Short words:", short)
print("Number of short words:", len(short))
print("Intermediate words:", intermediate)
print("Number of intermediate words:", len(intermediate))
print("Long words:", long)
print("Number of long words:", len(long))
```

2.10 Bookstore

```python
books = ["The Alchemist", "1984", "To Kill a Mockingbird", "The Great Gatsby", "Moby Dick"]
print("Original list of books:", books)
books = [book for book in books if book != "1984"]
print("After selling '1984':", books)
books.append("Pride and Prejudice")
books.append("The Catcher in the Rye")
print("After adding new books:", books)
print("Total number of books:", len(books))
print("First book:", books[0])
print("Last book:", books[len(books) - 1])
books.sort()
print("Books sorted alphabetically:", books)
count = 0
for book in books:
    if book == "Moby Dick":
        count = count + 1
if count > 0:
    print("Yes, 'Moby Dick' is available in the store.")
else:
    print("No, 'Moby Dick' is not available in the store.")
```
