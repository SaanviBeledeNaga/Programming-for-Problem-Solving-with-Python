3.1 Two Sum

```python
nums = [2, 7, 11, 15]
target = 9

for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        if nums[i] + nums[j] == target:
            print(i, j)
```

3.2 Contains Duplicate

```python
nums = [1, 2, 3, 1]
found = False
for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        if nums[i] == nums[j]:
            found = True
print(found)
```

3.3 Roman to Integer

```python
s = "LVIII"
values = {'I':1,'V':5,'X':10,'L':50,'C':100,'D':500,'M':1000}
total = 0
for i in range(len(s)):
    if i+1 < len(s) and values[s[i]] < values[s[i+1]]:
        total -= values[s[i]]
    else:
        total += values[s[i]]
print(total)
```

3.4 Plus One

```python
digits = [1, 2, 3]
i = len(digits) - 1
while i >= 0:
    if digits[i] < 9:
        digits[i] += 1
        break
    digits[i] = 0
    i -= 1
if i < 0:
    digits.insert(0, 1)
print(digits)
```

3.5 Majority Element

```python
nums = [2, 2, 1, 1, 1, 2, 2]
count = 0
candidate = None
for num in nums:
    if count == 0:
        candidate = num
    if num == candidate:
        count += 1
    else:
        count -= 1
print(candidate)
```

3.6 Richest Customer Wealth

```python
accounts = [[1, 5], [7, 3], [3, 5]]
max_wealth = 0
for customer in accounts:
    total = 0
    for money in customer:
        total += money
    if total > max_wealth:
        max_wealth = total
print(max_wealth)
```

3.7 Fizz Buzz

```python
n = 15
for i in range(1, n+1):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

3.8 Steps to Reduce Number to Zero

```python
num = 14
steps = 0
while num > 0:
    if num % 2 == 0:
        num = num // 2
    else:
        num -= 1
    steps += 1
print(steps)
```

3.9 Running Sum of 1D Array

```python
nums = [1, 2, 3, 4]
for i in range(1, len(nums)):
    nums[i] = nums[i] + nums[i-1]
print(nums)
```

3.10 Remove Element

```python
nums = [0,1,2,2,3,0,4,2]
val = 2
k = 0
for i in range(len(nums)):
    if nums[i] != val:
        nums[k] = nums[i]
        k += 1
print(k)
print(nums[:k])
```
