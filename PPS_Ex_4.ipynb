4.1 Student Report Card

```python
n = int(input())
data = {}
for x in range(n):
    name, subject, marks = input().split()
    marks = int(marks)
    if name not in data:
        data[name] = []
    data[name].append(marks)
averages = {}
for name in data:
    avg = sum(data[name]) / len(data[name])
    averages[name] = avg
for name in sorted(averages):
    print(f"{name}: {averages[name]}")
topper = max(averages, key=averages.get)
print(f"Topper: {topper} with average {averages[topper]}")
```

4.2 Grocery Inventory Tracker

```python
n = int(input())
inventory = {}
for _ in range(n):
    item, price, qty = input().split()
    inventory[item] = [int(price), int(qty)]
m = int(input())
total_bill = 0
for _ in range(m):
    item, buy_qty = input().split()
    buy_qty = int(buy_qty)
    if item in inventory and buy_qty <= inventory[item][1]:
        total_bill += buy_qty * inventory[item][0]
        inventory[item][1] -= buy_qty
print(f"Total Bill: {total_bill}")
print("Updated Inventory:")
for item in inventory:
    print(item, inventory[item][0], inventory[item][1])
```

4.4 Library Book Tracker

```python
n = int(input())
books = []
for _ in range(n):
    title, author, year, copies = input().split()
    books.append((title, author, int(year), int(copies)))
start_year, end_year = map(int, input().split())
found = False
for book in books:
    if start_year <= book[2] <= end_year:
        print(book)
        found = True
if not found:
    print("No books found in the given range.")
```

4.5 Banking Transaction Log

```python
n = int(input())
if n == 0:
    print("No transactions found.")
else:
    accounts = {}
    for _ in range(n):
        acc, ttype, amt = input().split()
        acc = int(acc)
        amt = float(amt)
        if acc not in accounts:
            accounts[acc] = 0
        if ttype == "credit":
            accounts[acc] += amt
        else:
            accounts[acc] -= amt
    for acc in sorted(accounts):
        print((acc, accounts[acc]))
```
