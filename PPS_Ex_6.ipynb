6.1 Count the Number of Vowels

```python
vowels = {'a','e','i','o','u'}
word = input()
original = word
word = word.lower()
count = 0
for ch in word:
    if ch in vowels:
        count += 1
print(f"The word '{original}' contains {count} vowel(s).")
```

6.2 Common Letters between Two Words

```python
word1 = input()
word2 = input()
set1 = set(word1.lower())
set2 = set(word2.lower())
common = set1 & set2
if common:
    print(f"The common letters between '{word1}' and '{word2}' are: {set(sorted(common))}")
else:
    print(f"The words '{word1}' and '{word2}' have no letters in common.")
```

6.3 Missing Letters Finder

```python
import string
word1 = input()
word2 = input()
alphabet = set(string.ascii_lowercase)
letters = set(word1.lower()) | set(word2.lower())
missing = sorted(alphabet - letters)
if not missing:
    print("All letters of the alphabet are present in the given words.")
else:
    print(f"Letters not present in either '{word1}' or '{word2}' are: {missing}")
```

6.4 City Travel Tracker

```python
a = set(map(str.strip, input().split(',')))
b = set(map(str.strip, input().split(',')))
print("Cities visited by both:", a & b)
print("Cities unique to Person A:", a - b)
print("Cities unique to Person B:", b - a)
print("All cities visited:", a | b)
```

6.5 DNA Sequence Comparison

```python
seq1 = input().strip()
seq2 = input().strip()
set1 = set(seq1)
set2 = set(seq2)
all_n = {'A','T','C','G'}
print("Common nucleotides:", set1 & set2)
print("Unique to first sequence:", set1 - set2)
print("Unique to second sequence:", set2 - set1)
print("Missing from first sequence:", all_n - set1)
print("Missing from second sequence:", all_n - set2)
```

6.6 Pangram Detector

```python
import string
sentence = input()
letters = set(ch.lower() for ch in sentence if ch.isalpha())
if letters == set(string.ascii_lowercase):
    print("The sentence is a pangram.")
else:
    print("The sentence is not a pangram.")
```

6.7 Shopping List Optimizer

```python
home = set(input().split())
office = set(input().split())
print("Items to buy for both:", home & office)
print("Items only for home:", home - office)
print("Items only for office:", office - home)
print("All unique items:", home | office)
```

6.8 Unique Words in Two Articles

```python
import string
a1 = input()
a2 = input()
translator = str.maketrans('', '', string.punctuation)
set1 = set(a1.lower().translate(translator).split())
set2 = set(a2.lower().translate(translator).split())
print("Common words:", set1 & set2)
print("Unique to first:", set1 - set2)
print("Unique to second:", set2 - set1)
print("Total distinct words:", len(set1 | set2))
```

6.9 College Course Enrollment

```python
a = set(input().split())
b = set(input().split())
c = set(input().split())
all_three = a & b & c
exact_two = (a & b) | (b & c) | (a & c)
exact_two = exact_two - all_three
only_one = (a | b | c) - exact_two - all_three
print("Students in all three courses:", all_three)
print("Students in exactly two courses:", exact_two)
print("Students in only one course:", only_one)
```

6.10 Lucky Draw Participants

```python
a = set(input().split())
b = set(input().split())
print("Participants in both events:", a & b)
print("Participants in only one event:", a ^ b)
```
