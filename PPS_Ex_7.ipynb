## 7.1 Sum of Natural Numbers and Their Cubes

```python
def sumN(n):
    return n * (n + 1) // 2
def sumNCubes(n):
    return (n * (n + 1) // 2)**2
n = int(input("Enter a positive integer n: "))
print(f"Sum of first {n} natural numbers: {sumN(n)}")
print(f"Sum of cubes of first {n} natural numbers: {sumNCubes(n)}")
```

## 7.2 Compute nth Fibonacci Number

```python
def fibonacci(n):
    if n < 0:
        return "Invalid input: n must be a non-negative integer."
    elif n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        a, b = 0, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        return b
n = int(input("Enter a non-negative integer n: "))
print(f"The {n}th Fibonacci number is: {fibonacci(n)}")
```

## 7.3 Next Guess for Square Root Approximation

```python
def nextGuess(guess, x):
    return (guess + x / guess) / 2
guess = float(input("Enter your current guess: "))
x = float(input("Enter the number x: "))
improved_guess = nextGuess(guess, x)
print(f"Next guess: {improved_guess:.2f}")
```

## 7.4 Letter Grade from Score

```python
def grade(score):
    if not (0 <= score <= 100):
        return "Invalid score"
    elif score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
score_input = float(input("Enter a score: "))
print(f"Grade: {grade(score_input)}")
```

## 7.5 Convert List of Number Strings to Numeric Values

```python
def toNumbers(strList):
    for i in range(len(strList)):
            strList[i] = float(strList[i])
input_str = input("Enter space-separated numeric values (e.g., 12 3.5 9 42): ")
str_list = input_str.split()
toNumbers(str_list)
print(str_list)
```

## 7.6 Guess the Number Game

```python
import random
def guess_the_number():
    secret_number = random.randint(0, 10)
    attempts = 0
    print("I'm thinking of a number between 0 and 10...")
    while True:
        guess = int(input("Enter your guess: "))
        attempts += 1
        if guess < secret_number:
            print("Too low! Try again.")
        elif guess > secret_number:
            print("Too high! Try again.")
        else:
            print(f"Congratulations! You guessed it in {attempts} attempts!")
            break
guess_the_number()
```

## 7.7 Playing with Numbers

```python
numbers = [7, 9, 15, 19, 24, 30, 37, 45, 50]
def divisible_by_3_and_5(num_list):
    result = [num for num in num_list if num % 3 == 0 and num % 5 == 0]
    print(f"Numbers divisible by both 3 and 5: {result}")
def divisible_by_3_or_5(num_list):
    result = [num for num in num_list if num % 3 == 0 or num % 5 == 0]
    print(f"Numbers divisible by 3 or 5: {result}")
def divisible_by_3_but_not_5_using_not(num_list):
    result = [num for num in num_list if num % 3 == 0 and not (num % 5 == 0)]
    print(f"Numbers divisible by 3 but not 5 (using not): {result}")
def divisible_by_3_but_not_5_without_not(num_list):
    result = [num for num in num_list if num % 3 == 0 and num % 5 != 0]
    print(f"Numbers divisible by 3 but not 5 (without using not): {result}")
divisible_by_3_and_5(numbers)
divisible_by_3_or_5(numbers)
divisible_by_3_but_not_5_using_not(numbers)
divisible_by_3_but_not_5_without_not(numbers)
```

## 7.8 Flipping Coins – Counting Heads and Tails

```python
import random
def simulate_flips(num_flips):
    outcomes = [random.choice([True, False]) for _ in range(num_flips)]
    heads = sum(outcomes)
    tails = num_flips - heads
    return outcomes, heads, tails
def calculate_ratio(heads, tails):
    if tails == 0:
        return float('inf')
    return heads / tails
print("Flipping the coin 8 times...")
outcomes_8, heads_8, tails_8 = simulate_flips(8)
print(f"Outcomes: {outcomes_8}")
print(f"Heads: {heads_8} | Tails: {tails_8}")
ratio_8 = calculate_ratio(heads_8, tails_8)
print(f"Ratio (Heads/Tails): {ratio_8:.2f}")
print("\nFlipping the coin 1000 times...")
outcomes_1000, heads_1000, tails_1000 = simulate_flips(1000)
print(f"Heads: {heads_1000} | Tails: {tails_1000}")
ratio_1000 = calculate_ratio(heads_1000, tails_1000)
print(f"Ratio (Heads/Tails): {ratio_1000:.2f}")
print("\nObservation: As the number of flips increases, the ratio of heads to tails becomes closer to 1.")
```

## 7.9 Comparator

```python
def compare(num1, num2):
    if num1 > num2:
        return num1
    else:
        return num2
number1 = float(input("Enter the first number: "))
number2 = float(input("Enter the second number: "))
larger_number = compare(number1, number2)
print(f"The larger number is: {larger_number}")
```

## 7.10 Digit Power Sum

```python
def digit_power_sum(number, power):
    s = str(number)
    total_sum = 0
    for digit_char in s:
        digit = int(digit_char)
        total_sum += (digit ** power)
    return total_sum
num_input = int(input("Enter a number: "))
power_input = int(input("Enter the power to raise each digit to: "))
result = digit_power_sum(num_input, power_input)
print(f"The sum of the digits of {num_input} raised to the power {power_input} is: {result}")
```
