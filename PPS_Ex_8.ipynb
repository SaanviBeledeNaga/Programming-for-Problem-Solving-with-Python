## 8.1 Power and Energy Calculator

```python
V, I = map(int, input().split())
t = int(input())

def calculate_power_energy(voltage, current, time_hours):
    power = voltage * current
    energy = power * time_hours
    return power, energy

power_val, energy_val = calculate_power_energy(V, I, t)

print(f"Power: {power_val} W")
print(f"Energy: {energy_val} Wh")
```

## 8.2 Interest Analyzer

```python
import math

def simple_interest(p, r, t):
    return (p * r * t) / 100

def compound_interest(p, r, t):
    rate_decimal = r / 100
    return p * (math.pow((1 + rate_decimal), t)) - p

P = int(input())
R = float(input())
T = int(input())

si = simple_interest(P, R, T)
ci = compound_interest(P, R, T)

print(f"Simple Interest: {si}")
print(f"Compound Interest: {ci}")
```

## 8.3 Statistician’s Data Analyzer

```python
from collections import Counter

def mean(data):
    return sum(data) / len(data)

def median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    if n % 2 == 1:
        return sorted_data[n // 2]
    else:
        mid1 = sorted_data[n // 2 - 1]
        mid2 = sorted_data[n // 2]
        return (mid1 + mid2) / 2

def mode(data):
    if not data:
        return None
    counts = Counter(data)
    max_count = max(counts.values())
    modes = [key for key, value in counts.items() if value == max_count]
    return min(modes)

n = int(input())
data_values = list(map(int, input().split()))

mean_val = mean(data_values)
median_val = median(data_values)
mode_val = mode(data_values)

print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Mode: {mode_val}")
```

## 8.4 Astronomer’s Orbit Calculator

```python
import math

def orbit_length(r):
    return 2 * math.pi * r

def orbit_velocity(r, T):
    circumference = orbit_length(r)
    return circumference / T

r = float(input())
T = float(input())

circumference = orbit_length(r)
velocity = orbit_velocity(r, T)

print(f"Orbital Circumference: {circumference:.2f} km")
print(f"Orbital Velocity: {velocity:.2f} km/s")
```

## 8.5 Farmer’s Field Division

```python
def number_of_plots(length, width, side):
    plots_along_length = length // side
    plots_along_width = width // side
    return plots_along_length * plots_along_width

def remaining_area(length, width, side):
    total_field_area = length * width
    num_plots = number_of_plots(length, width, side)
    covered_area = num_plots * (side * side)
    return total_field_area - covered_area

length = int(input())
width = int(input())
side = int(input())

num_plots = number_of_plots(length, width, side)
rem_area = remaining_area(length, width, side)

print(f"Number of plots: {num_plots}")
print(f"Remaining area: {rem_area} sq.m")
```

## 8.6 Banker’s Loan Evaluator

```python
def evaluate_loan(annual_income, credit_score, existing_loan_amount):
    status = "Rejected"
    max_eligible_loan = annual_income * 0.5

    if credit_score >= 750 and annual_income >= 500000:
        status = "Approved"
    elif 650 <= credit_score <= 749 and annual_income >= 300000:
        if existing_loan_amount <= 200000:
            status = "Conditionally Approved"
        else:
            status = "Rejected"
    elif credit_score < 650:
        status = "Rejected"

    return status, max_eligible_loan

annual_income = int(input())
credit_score = int(input())
existing_loan_amount = int(input())

loan_status, max_loan_amount = evaluate_loan(annual_income, credit_score, existing_loan_amount)

print(loan_status)
print(f"{max_loan_amount}")
```
